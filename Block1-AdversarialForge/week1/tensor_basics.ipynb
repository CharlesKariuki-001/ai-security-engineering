```
# 1. Creating Tensors
# Scalar (0D)
scalar = torch.tensor(7)
print("Scalar:", scalar)

# Vector (1D)
vector = torch.tensor([1, 2, 3, 4])
print("Vector:", vector)

# Matrix (2D)
matrix = torch.tensor([[1, 2], [3, 4]])
print("Matrix:", matrix)

# 3D Tensor
tensor_3d = torch.rand(3, 4, 5)   # random values
print("3D Tensor shape:", tensor_3d.shape)
```

```
# 2. Important Operations
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

print("Addition:", a + b)
print("Multiplication (element-wise):", a * b)
print("Matrix Multiplication:", torch.matmul(a, b))
```
